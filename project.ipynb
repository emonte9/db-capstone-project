```python
import mysql.connector as connector
```

```python
connection = connector.connect(user = "root", password = "", db="LittleLemon2DB")
```

```python
# cursor = connection.cursor()
cursor = connection.cursor(buffered=True)
```

```python
cursor.execute("SHOW TABLES")
for table in cursor:
    print(table)
```

```python
query = """
Select c.FullName, c.ContactNumber, c.Email, o.TotalCost
FROM Orders o
JOIN Customers c ON  o.CustomerID = c.CustomerID
WHERE o.TotalCost > 60
"""

# cursor.fetchall()   # clears unread results
cursor.execute(query)
results = cursor.fetchall()

cols = cursor.column_names

# Print column names and records from results using for loop
print("""Data in the " table:""")
print(cols)
for result in results:
    print(result)


```
